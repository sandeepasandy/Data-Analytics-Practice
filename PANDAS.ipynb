```
#filtering
import pandas as pd
data = {
    "Name":["Ravi","priya","Arun","sneha","kiran","Anjali","Vikram","Meena"],
    "Age" :[22,24,None,25,None,26,20,27],
    "City":["Hyderabad","chennai","Bangalore","Hyderabad","Mumbai",None,"Bangalore","Mumbai"],
    "Salary":[30000,40000,28000,50000,35000,45000,None,55000]
}
df = pd.DataFrame(data)
print(df)
df.iloc[0:3]
df.loc[0]
print("___________________")#getting the members whose age is above 21 with the boolean values
e = df["Age"]>21
print(e)
print("___________________")
d = df[df["Age"]>21]
print(d)
print("___________________")#getting the salary whose salary is greater than 45000
s = df[df["Salary"]>45000]
print(s)
print("_________________")#getting the members whose age is greater than 22 and city is equal to Hyderabad
g = df[                      #by using and operator
    (df["Age"]>22) &
     (df["City"]=="Hyderabad")
]
print(g)
print("_________________")#getting the members whose salary is greater than 45000 and city is equal to mumbai
e = df[                   #with using or(|) operator
    (df["Salary"]>45000) |
    (df["City"]=="Mumbai")
]
print(e)
df.isna()
df.dropna(axis=0)#deletes the rows which have none values
df.dropna(how='all')
df.dropna(axis=0,inplace=True)#changes the  null values permantely
print(df)
df.fillna({
    'Age':0,
    'City':'Unknown'
})
print(df)
```

```
df = pd.DataFrame(data)
df.fillna({
    'Age':0,
    'City':'Unknown'
})
print(df)
df['Age'] = df['Age'].fillna(#Gives the mean values of the null values
    df['Age'].mean())
print(df)
df['Salary'] =df['Salary'].fillna(
    df['Salary'].mean())
print(df)
```

```
import pandas as pd
emp_details ={
    "EmpID":[101,102,103,104,101],
    "City":["Hyderabad","Chennai","Bangalore",None,"Hyderabad"],
    "Experience":[None,4,3,5,2],
    "Age":[25,30,None,35,25]
}
df = pd.DataFrame(emp_details)
print(df)
df.isna()
df.isna().sum()
df.fillna({
    'Experience':0,
    'City':'Unknown'
})
df['Age'] = df['Age'].fillna(
    df['Age'].median())
print(df)
df.rename(
    columns={
        'EmpID':'Employee_id'
    }
)
df["Salary"]=[20000,35000,50000,1000,None]
print(df)
df["Age"] = df["Age"]+1#getting age of employee after one year-updating a column
print(df)
df["Bonus"]=df["Salary"]*0.10#getting bonus of every employeee
print(df)
df.drop(#drop can remove columns from a DataFrame
    'Age',
    axis=1,
    inplace=True,#it is used to delete permanantely
)
df.drop(#used to delete multiple columns
    columns=['Salary','Bonus'],
)
print("__________________________")
df.loc[len(df)]=[106,"Vizag",3,25,45000]#adding a new row
print(df)
df.drop(2,axis=0,inplace=True)#Deleting a row
print(df)
```
