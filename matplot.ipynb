```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load data
df = sns.load_dataset('titanic')


# 1. Count Plot
sns.countplot(
    data=df,
    x='class',
    hue='sex'
)
plt.title('Passengers by Class and Sex')
plt.show()
```

```python
# 2. Box Plot
sns.boxplot(
    data=df,
    x='class',
    y='age'
)
plt.title('Age Distribution by Class')
plt.show()
```

```python
# 3. Violin Plot
sns.violinplot(
    data=df,
    x='class',
    y='age',
    hue='sex'
)
plt.title('Age Distribution by Class and Sex')
plt.show()
```

```python

# 4. Density Plot
sns.kdeplot(
    data=df,
    x='age',
    hue='sex',
    fill=True
)
plt.title('Age Density by Sex')
plt.show()
```

```python
# 5. Regression Plot
sns.regplot(
    data=df,
    x='age',
    y='fare'
)
plt.title('Age vs Fare')
plt.show()
```

```python

# 6. Joint Plot
sns.jointplot(
    data=df,
    x='age',
    y='fare',
    kind='reg'
)
plt.show()
```

```python
# 7. Pair Plot
sns.pairplot(
    df[['age', 'fare', 'sibsp', 'parch']],
    corner=True
)
plt.show()
```

```python
# 8. Correlation Heatmap
corr = df[['age', 'fare', 'sibsp', 'parch']].corr()

sns.heatmap(
    corr,
    annot=True
)

plt.title('Correlation Heatmap')
plt.show()
```

```python
# ==========================================
# PLOTLY COMPLETE PRACTICE
# ==========================================

# 1. IMPORT LIBRARIES
import numpy as np
import seaborn as sns
import plotly.express as px
import plotly.graph_objects as go


# ==========================================
# 2. LOAD TITANIC DATA
# ==========================================

df = sns.load_dataset("titanic")

print(df.head())

```

```python
# ==========================================
# 3. BASIC PLOTLY LINE
# ==========================================

fig = px.line(
    x=[1, 2, 3, 4, 5],
    y=[10, 20, 15, 30, 25]
)

fig.show()

```

```python
# ==========================================
# 4. CONVEX 3D GRAPH - SPHERE
# ==========================================

u = np.linspace(0, 2 * np.pi, 50)
v = np.linspace(0, np.pi, 50)

x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))

fig = go.Figure(
    data=[
        go.Surface(
            x=x,
            y=y,
            z=z
        )
    ]
)

fig.show()
```

```python
# ==========================================
# 5. NON-CONVEX 3D GRAPH
# ==========================================

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = np.sin(np.sqrt(X**2 + Y**2))

fig = go.Figure(
    data=[
        go.Surface(
            x=X,
            y=Y,
            z=Z
        )
    ]
)

fig.show()
```
