```python
#29-08-2026
import numpy as np
#line spacing
arr = np.linspace(0,10,3)
print(arr)
#Array shape
arr = np.array([10,20,30,40])
print(arr.shape)
#array reshape
arr = np.array([10,20,30,40])
print(arr.reshape)
#ravel
arr = np.arange(1,11).reshape(2,5,1)
print(arr)
r = arr.ravel()
print(r)
#Transpose the array
#rows to columns and columns into rows
arr = np.arange(1,11).reshape(2,5,1)
print(arr)
r = arr.T
print(r)
r = arr.ravel()
print(r)
#Data type conversion
#can change the numbers into float or any values
arr = np.array([10,20,30])
new_arr = arr.astype(float)
print(new_arr)
print(new_arr.dtype)
#Indexing
arr = np.array([10,20,30,40,50])
print(arr[0:2])#for getting the first two numbers
print(arr[-2:])#for getting the last two numbers
print(arr[-3:-1])#for getting the middle elements
#indexing in 2-Dimension
arr = np.arange(1,11).reshape(2,5)
print(arr)
print(arr[0,3])
#slicing
#if we mention as 0:1 if there are two rows then it will print only one row
arr = np.arange(1,11).reshape(2,5)
print(arr)
print(arr[1,4])
print(arr[0:1])
print(arr[0:2, 0:4])#slicing in two dimensions
#Modification in single dimension
#whreas here we have given the index5 it will change it into 100
arr = np.arange(10,20)
print(arr)
arr[5]=100
print(arr)
#2D
arr[1:4]=80
print(arr)
#insert
n = np.insert(arr,3,89)
print(n)
#for inserting mutiple values
arr = np.array([10,20,30,40,50])
new_arr = np.insert(arr,2,[100,200])
print(new_arr)
#Delete
#for deleting one element
arr = np.array([10,20,30,40,50])
new_arr = np.delete(arr,2)
print(new_arr)
#for deleting multiple elements
arr = np.array([10,20,30,40,50])
new_arr = np.delete(arr,[2,3])
print(new_arr)
```

```python
#01-09-26
#aggregation functions in numpy
import numpy as np
#1D
arr = np.array([10,20,30,40,50])
print(np.sum(arr))
#2D
arr = np.array([10,20,30,40,50,60]).reshape(2,3)
print(np.sum(arr))
#min
print(np.min(arr))
#max
print(np.max(arr))
#sum with axis
arr = np.array([10,20,30,40,50,60]).reshape(2,3)
print(arr)
print(np.sum(arr,axis=0))
#mean
#1D
arr = np.array([10,20,30,40,50])
print(np.mean(arr))
#2D
arr = np.array([10,20,30,40,50,60]).reshape(2,3)
print(np.mean(arr))
#Variance
#1D
arr = np.array([10,20,30,40,50])
print(np.var(arr))
#2D
arr = np.array([10,20,30,40,50,60]).reshape(2,3)
print(np.var(arr))
#standard deviation(std)
#1D
arr = np.array([10,20,30,40,50])
print(np.std(arr))
#2D
arr = np.array([10,20,30,40,50,60]).reshape(2,3)
print(np.std(arr))
#percentile:
#0 to 100
#if we give 100 means it will return the max value
#if we give any another number it will give another number
arr = np.array([10,20,30,40,50])
print(np.percentile(arr,50))
#quantile:
#0 to 1 like 0.20,0.30 etc...
np.quantile(arr,0.50)
#argmax
#returns the position of the element in array which is in the highest value
#1D
arr = np.array([10,90,30,70,40])
print(np.argmax(arr))
#2D
arr = np.array([10,90,30,70,40,10]).reshape(2,3)
print(np.argmax(arr))
#argmin
#1D
arr = np.array([10,90,30,70,40])
print(np.argmin(arr))
#2D
#argmin
arr = np.array([10,90,30,70,40,10]).reshape(2,3)
print(np.argmin(arr))
#log10()
#1D
arr = np.array([10,100,1000])
print(np.log10(arr))
#2D
arr = np.array([10,90,30,70,40,60]).reshape(2,3)
print(np.log10(arr))
#log2()
arr = np.array([10,90,30,70,40,60]).reshape(2,3)
print(np.log10(arr))
print(np.log(arr))
print(np.log(2))
print(np.log(3))
#round
arr = np.array([10.25, 20.67,30.93])
print(np.round(arr))
```

```python
#round
arr = np.array([10.253, 20.675,30.939])
print(np.round(arr))
print(np.round(arr,2))
#floor
#adjusts to lower limit
arr = np.array([10.2,10.8,20.9])
print(np.floor(arr))
#ceil()
#adjusts to upper limit
arr = np.array([10.2,10.8,20.9])
print(np.ceil(arr))
#abs
#if we given negative numbers also it will only return the positive numbers
arr = np.array([10,-5,6,-19])
print(np.abs(arr))
#sort
#ascending order
arr = np.array([90,120,150,40,10])
result = np.sort(arr)
print(result)
#changing to descending order
arr = np.array([90,120,150,40,10])
result = np.sort(arr)[::-1]
print(result)
```

```python
#04-09-26
#split 1D array
import numpy as np
arr = np.array([10,20,30,40,50,60])
result = np.split(arr,3)#splits the elements in 3 equal parts
print(result)
arr = np.array([10,20,30,40,50,60])
result = np.split(arr,2)#splits the elements in 2 equal parts
print(result)
#customized splitting
#passing the values in the list
arr = np.array([10,20,30,40,50,60])
result = np.split(arr,[2,4])
print(result)
#split on 2D array
arr = np.array([
    [10,20],
    [30,40],
    [50,60],
    [70,80]
])
result = np.split(arr,2,axis=0)#for rows if we want to do as columns then axis=1
print(result)
#customized splitting on 2d array
arr = np.array([
    [10,20,30],
    [30,40,50],
    [50,60,70],
    [70,80,90]
])
result = np.split(arr,[1,3],axis=1)
print(result)
```
