```python
import plotly.express as px
```

```python
#line plot in plotly
x=[1,2,3,4,5]
y=[10,20,15,35,17]
fig = px.line(
    x=x,
    y=y
)
fig.show()
```

```python
#3D Scatter plot
fig =px.scatter_3d(
    x=[1,2,3,4,5],
    y=[12,23,4,7,12],
    z=[22,11,77,34,28]
)
fig.show()
```

```python
import seaborn as sns
df = sns.load_dataset('titanic')
print(df)
```

```python
from matplotlib import markers
import plotly.express as px
px.scatter_3d(df,x="pclass",y="fare",z="sibsp")
```
