```python
df = sns.load_dataset('titanic')
print(df)
#lineplot in titanic
sns.lineplot(data=df,x="pclass",y="fare",hue="sex",marker="*")
plt.title("Scatters")
print(df)
```

```python
sns.barplot(data=df,x="survived",y="sex")
plt.title("Survived rate")
print(df)
```

```python
sns.scatterplot(data=df,x="survived",y="pclass",marker="*")
plt.title("Survival")
print(df)
```

```python
sns.boxplot(data=df,x="survived",y="age")
plt.title("Box plot")
print(df)
```

```python
sns.violinplot(data=df,x="survived",y="age")
plt.title("violin chart")
print(df)
```

```python
#Heatmap in seaborn
corr = df[['age', 'fare','sibsp','parch']].corr()
sns.heatmap(
  corr,
  annot=True
)
plt.show()
```

```python
#for knowing the count by the barchart.
sns.countplot(
    data=df,
    x='class',
    hue="sex"
)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('titanic')
sns.pairplot(
    df[['age','fare','sibsp','parch']]
)
plt.show()
```

```python
#heatmap function by pivot table
pivot = pd.pivot_table(
    df,
    values='survived',
    index='sex',
    columns='class',
    aggfunc='mean'
)
print(pivot)
sns.heatmap(
    pivot,annot=True,fmt='.2f')
```

```python
#pivot table using 2 indexing.
pivot = pd.pivot_table(
    df,
    values='survived',
    index=['sex','class'],
    aggfunc='mean'
)
print(pivot)
sns.heatmap(
    pivot,annot=True,fmt='.2f')
```

```python
#Regression plot:The line covers all the points which are satisfied
sns.regplot(
    data=df,
    x='age',
    y='fare'
)
plt.show()
```

```python
#Density plot- This will covers where the points are more
sns.kdeplot(
    data=df,
    x='pclass',
    fill=True
)
plt.show()
```

```python
#FacetGrid presents only the grid then we can make some charts in that grid
g = sns.FacetGrid(
    df,
    col='pclass'
)
g.map(
    sns.histplot,
    'age'
)
plt.show()
```

```python
#jointplot:It is nothing but combination of scatter and hist.
sns.jointplot(
    data=df,
    x='age',
    y='fare'
)
plt.show()
```
